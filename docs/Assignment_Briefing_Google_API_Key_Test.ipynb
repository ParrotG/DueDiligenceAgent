### 1.4: Select the Gemini models

- [Gemini 3.5 Flash-Lite](https://docs.cloud.google.com/gemini-enterprise-agent-platform/models/gemini/3-5-flash-lite) (gemini-3.5-flash-lite)
> It is the latest in our cost-effective Flash-Lite line of models. It's optimized for simple coding tasks, precise document understanding, and lightweight agentic workflows that require fast inference at minimal cost.

- [Gemini 3.1 Flash-Lite](https://docs.cloud.google.com/gemini-enterprise-agent-platform/models/gemini/3-1-flash-lite) (gemini-3.1-flash-lite)
> It is our most cost-efficient Gemini model, optimized for low latency use cases for high-volume, cost-sensitive LLM traffic.

````python
# --- Define Model Constants for easier use ---

# More supported models can be referenced here: https://ai.google.dev/gemini-api/docs/models#model-variations
# https://docs.cloud.google.com/gemini-enterprise-agent-platform/models/gemini/3-1-flash-lite
# https://docs.cloud.google.com/gemini-enterprise-agent-platform/models/model-versions
MODEL_GEMINI_3_1_FLASH_LITE = "gemini-3.1-flash-lite"
MODEL_GEMINI_3_5_FLASH_LITE = "gemini-3.5-flash-lite"

print("\nEnvironment configured and model selected.")
````

````python
client = genai.Client(api_key=GOOGLE_API_KEY)

# List all the supported models
for model in client.models.list():
    print(model.name)
````

````python
# Constructing prompt with query

query = "Describe the NVIDIA GB200 architecture?" # "How to craft effective prompts?"

# This prompt is where you can specify any guidance on tone, or what topics the model should stick to, or avoid.
prompt = f"""You are a NVIDIA GPU architect with deep knowledge and understanding in accelerated computing and GPU architecture.
Organise your thoughts and describe the concepts clearly to the user without much experience.
Use clear paragraphs and formatting. If you not know the answer, just say you do not know.

QUESTION: {query}
"""

prompt
````

````python
answer = client.models.generate_content(
                                        model=MODEL_GEMINI_3_1_FLASH_LITE,
                                        contents=prompt
                                       )

Markdown(answer.text)
````
